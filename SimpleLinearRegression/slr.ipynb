#### Regression is finding the best fit line!!
 Difference between best fit line value (i.e. predicted) and actual value is `Residual Error`

### Maths to find Best Fit line:
- `y= mx+c` line (where m = slope or coefficient)
    - i.e. slope = 1 unit change in x = how much unit change in y?
    - slope = most imp stuff to change

- `Andrew NG notation` : H<sub>Θ</sub>(x) = Θ<sub>0</sub> + Θ<sub>1</sub>x<sub>1</sub> + Θ<sub>2</sub>x<sub>2</sub> +... Θ<sub>n</sub>x<sub>n</sub>
- `What we need` : H<sub>Θ</sub>(x) = Θ<sub>0</sub> + Θ<sub>1</sub>x<sub>1</sub> <-- Translated from y = mx+c
- {Y is a linear function of X}
