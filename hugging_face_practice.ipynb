# HuggingFacePipeline

## How to Access Hugging Face Models Locally with LangChain

Previously, we could use `HuggingFaceEndpoint` to access Hugging Face models through hosted inference APIs. However, Hugging Face Inference Providers no longer include free monthly inference credits for regular free-tier users, so using these hosted APIs may require purchasing credits or subscribing to a paid plan.

Therefore, we will now use `HuggingFacePipeline` to run Hugging Face models directly in our environment, such as Google Colab. This approach downloads the model weights from the Hugging Face Hub and runs the model using the available CPU or GPU resources.

Unlike `HuggingFaceEndpoint`, `HuggingFacePipeline` does not require paid inference API credits. It allows us to experiment with language models and use LangChain methods such as `llm.invoke()` without relying on a hosted inference service. However, we must ensure that our environment has sufficient memory and computational resources to run the selected model.

```python
from langchain_huggingface import HuggingFacePipeline #Import HuggingFace Pipeline
```

```python
llm = HuggingFacePipeline.from_model_id(
    model_id="HuggingFaceTB/SmolLM2-360M-Instruct",
    task="text-generation",
    pipeline_kwargs={
        "max_new_tokens": 100,
        "top_k": 50,
        "temperature": 0.1,
    },
) #Define The LLM pipeline
```

```python
llm.invoke("Hugging Face is")
```
